```python
# Program Menghitung Total Bayar Pembelian Barang

# Modifikasi 1: input via keyboard
nama_barang = input("Masukkan nama barang : ")
belanja = int(input("Masukkan total belanja (Rp) : "))

# Modifikasi 2: diskon 35% jika belanja di atas 1.000.000
diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon

print("\n===== STRUK PEMBELIAN =====")
print("Barang                   :", nama_barang)
print("Total belanja            : Rp", belanja : %.2f)
print("Diskon                   : Rp", diskon)
print("Total harga setelah diskon = Rp", total_harga)
```

```python
# Program Menghitung Nilai Akhir dan Status Kelulusan Mahasiswa

# Nilai mahasiswa terdiri dari 5 variabel, diinput via keyboard
tugas = float(input("Masukkan nilai tugas      : "))
kuis = float(input("Masukkan nilai kuis       : "))
praktikum = float(input("Masukkan nilai praktikum  : "))
uts = float(input("Masukkan nilai UTS         : "))
uas = float(input("Masukkan nilai UAS         : "))

# Bobot: tugas 20%, kuis 10%, praktikum 20%, UTS 20%, UAS 30%
nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)

# Standar kelulusan diinput via keyboard
standar_kelulusan = float(input("Masukkan standar kelulusan : "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan

# Hanya satu status yang ditampilkan
status = ("Perlu Remedial", "Lulus")[apakah_lulus]

print("\n===== HASIL PENILAIAN =====")
print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Status: {status}")
```

```python
# Input jumlah detik dari pengguna
try:
    total_detik = int(input("Masukkan jumlah detik: "))
    if total_detik < 0:
        print("Jumlah detik tidak boleh negatif.")
    else:
        # Hitung jumlah jam
        jam = total_detik // 3600  # 1 jam = 3600 detik
        sisa_detik_setelah_jam = total_detik % 3600

        # Hitung jumlah menit
        menit = sisa_detik_setelah_jam // 60  # 1 menit = 60 detik
        sisa_detik = sisa_detik_setelah_jam % 60

        # Print hasil konversi
        print(f"{total_detik} detik = {jam} jam, {menit} menit, {sisa_detik} detik")
except ValueError:
    print("Input tidak valid. Harap masukkan angka bulat.")
```

```python
# Soal 1: Belanja Buku

# Input via keyboard (poin 5)
jumlah_buku = int(input("Jumlah buku           : "))
harga_buku = int(input("Harga per buku (Rp)   : "))
persen_diskon = int(input("Diskon (%)            : "))
uang_budi = int(input("Uang Budi (Rp)        : "))
jumlah_orang = int(input("Jumlah orang patungan: "))

# Poin 1: total, diskon, bayar
total = jumlah_buku * harga_buku
diskon = total * persen_diskon // 100
bayar = total - diskon

# Poin 2: uang kembalian
kembalian = uang_budi - bayar

# Poin 3: patungan dalam bilangan bulat
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

# Poin 4: apakah uang cukup
uang_cukup = uang_budi >= bayar

print("\n==========")
print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
# Soal 2: PLTS Atap Rumah

# Input via keyboard
jumlah_panel = int(input("Jumlah panel surya          : "))
daya_panel = float(input("Daya per panel (Wp)         : "))
jam_efektif = float(input("Penyinaran efektif (jam/hari): "))
efisiensi = float(input("Efisiensi sistem (contoh 0.8): "))
kebutuhan = float(input("Kebutuhan listrik (kWh/hari) : "))
tarif = float(input("Tarif listrik (Rp/kWh)        : "))

# Poin 1: energi per hari (kWh)
energi_harian = jumlah_panel * daya_panel / 1000 * jam_efektif * efisiensi

# Poin 2: energi per bulan (30 hari) dan nilai rupiahnya
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif

# Poin 3: kekurangan energi per hari
kekurangan = kebutuhan - energi_harian

# Poin 4: apakah PLTS mencukupi
mencukupi = energi_harian >= kebutuhan

print("\n==========")
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  : %.2f" % nilai_listrik)
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
# Soal 3: Generator Diesel dan Rugi Kabel

# Input via keyboard (poin 6)
daya = float(input("Daya generator (kW)          : "))
jam = float(input("Lama operasi (jam/hari)     : "))
konsumsi = float(input("Konsumsi BBM (liter/kWh)    : "))
harga_solar = float(input("Harga solar (Rp/liter)      : "))
kapasitas_tangki = float(input("Kapasitas tangki (liter)    : "))
arus = float(input("Arus kabel (A)              : "))
hambatan = float(input("Hambatan kabel (ohm)        : "))

# Poin 1: energi per hari (kWh)
energi = daya * jam

# Poin 2: kebutuhan solar per hari dan biayanya
solar = energi * konsumsi
biaya = solar * harga_solar

# Poin 3: berapa hari tangki penuh cukup dan sisa solar
hari_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

# Poin 4: rugi daya kabel (P = I^2 x R)
rugi_kabel = arus ** 2 * hambatan

# Poin 5: rugi tinggi jika > 150 W dan arus minimal 20 A
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("\n==========")
print("Energi (kWh)     :", round(energi))
print("Solar (liter)    :", solar)
print("Biaya (Rp)       : %.2f" % biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
