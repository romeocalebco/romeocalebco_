```python
tahun = int(input("Input tahun: "))

if (tahun % 4) == 0:
    if (tahun % 100) == 0:
        if (tahun % 400) == 0:
            print("Tahun Kabisat")
        else:
            print("Bukan Tahun Kabisat")
    else:
        print("Tahun Kabisat")
else:
    print("Bukan Tahun Kabisat")
```

```python
# Data makanan
nama_makanan = "Nasi Goreng"  # string
jumlah_porsi = 3              # integer
harga_per_porsi = 15000.0     # float
kode_menu_hex = 255           # integer untuk hex
kode_menu_octal = 64          # integer untuk octal

# Menampilkan dengan format string
print("Nama Makanan: %s" % nama_makanan)       # %s untuk string
print("Jumlah Porsi: %d" % jumlah_porsi)       # %d untuk integer
print("Harga per Porsi: %f" % harga_per_porsi) # %f untuk float
print("Kode Menu (Hex): %x" % kode_menu_hex)   # %x untuk hexadecimal
print("Kode Menu (Octal): %o" % kode_menu_octal) # %o untuk octal
```

```python
# Input jumlah detik dari pengguna
try:
    total_detik = int(input("Masukkan jumlah detik: "))
    if total_detik < 0:
        print("Jumlah detik tidak boleh negatif.")
    else:
        # Hitung jumlah jam
        jam = total_detik // 3600  # 1 jam = 3600 detik
        sisa_detik_setelah_jam = total_detik % 3600

        # Hitung jumlah menit
        menit = sisa_detik_setelah_jam // 60  # 1 menit = 60 detik
        sisa_detik = sisa_detik_setelah_jam % 60

        # Print hasil konversi
        print(f"{total_detik} detik = {jam} jam, {menit} menit, {sisa_detik} detik")
except ValueError:
    print("Input tidak valid. Harap masukkan angka bulat.")
```
