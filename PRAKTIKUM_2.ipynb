```python
Tegangan_Harian = [217.5,219.0,220.3,218.7,221.1,216.9]
# A. Cetak pembacaan pertama dan terakhir pakai indeks
print("Pembacaan pertama",Tegangan_Harian[0])
print("Pembacaan Terakhir",Tegangan_Harian[-1])
# B. Tambahkan 1 pembacaan baru 22.0 dengan append()
Tegangan_Harian.append(22.0)
print("List setelah ditambah:", Tegangan_Harian)

# C. Cetak jumlah total data dengan len()
print("Jumlah total data:", len(Tegangan_Harian))
```

```python
# Dictionary spesifikasi motor listrik
motor = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 5.5,
    "tegangan_V": 380,
    "rpm": 1450
}

# Cetak salah satu nilai lewat key
print("RPM awal:", motor["rpm"])

# Ubah nilai rpm
motor["rpm"] = 1500

# Cetak ulang seluruh dictionary
print("Dictionary setelah diubah:", motor)
```

```python
# List of dictionary data pembangkit
pembangkit = [
    {"nama": "PLTU Unit 1", "kapasitas_MW": 600, "status": "Beroperasi"},
    {"nama": "PLTG Unit 2", "kapasitas_MW": 150, "status": "Standby"},
    {"nama": "PLTA Unit 3", "kapasitas_MW": 300, "status": "Beroperasi"}
]

# Cetak nama unit kedua
print("Unit kedua:", pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[1]["status"] = "Beroperasi"

# Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "PLTS Unit 4", "kapasitas_MW": 50, "status": "Konstruksi"})

# Cetak seluruh data
print("\nData seluruh pembangkit:")
for unit in pembangkit:
    print(unit)
```
