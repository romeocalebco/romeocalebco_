```python
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i in range(len(data)):
    n = i + 1
    try:
        nilai = float(data[i])
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

    if nilai > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")
    jumlah_valid += 1

print(f"Jumlah data valid: {jumlah_valid}")
```

```python
# ============================================================
# MONITOR BEBAN GENSET PLTD
# Masalah   : operator perlu memantau beban genset tiap 3 jam,
#             mendeteksi beban berlebih, dan memperkirakan
#             berapa lama BBM di tangki masih cukup.
# Cara kerja: data (teks, ada sensor rusak) dibaca satu per satu
#             dengan for + try-except, diklasifikasi dengan
#             if-elif-else, lalu hasilnya dihitung dan
#             ketahanan BBM disimulasikan dengan while.
# ============================================================

KAPASITAS_KW = 400
KONSUMSI_LITER_PER_KWH = 0.28
TANGKI_LITER = 500

jam_ukur = [0, 3, 6, 9, 12, 15, 18, 21]
data_beban = ["180", "150", "210", "340", "385", "ERR", "395", "260"]

beban_valid = []
jam_valid = []
sensor_rusak = 0
jumlah_kritis = 0

print("=== LAPORAN BEBAN GENSET ===")
for i in range(len(data_beban)):
    jam = jam_ukur[i]

    # Sensor rusak dilewati, program tetap jalan
    try:
        beban = float(data_beban[i])
    except ValueError:
        print(f"Jam {jam:02d}:00 | sensor rusak, data dilewati")
        sensor_rusak += 1
        continue

    # Klasifikasi berdasarkan persen kapasitas
    persen = beban / KAPASITAS_KW * 100
    if persen >= 95:
        status = "KRITIS"
        jumlah_kritis += 1
    elif persen >= 80:
        status = "TINGGI"
    else:
        status = "NORMAL"

    bar = "█" * int(persen // 10)
    print(f"Jam {jam:02d}:00 | {beban:5.0f} kW | {bar:<10} {persen:5.1f}% | {status}")

    beban_valid.append(beban)
    jam_valid.append(jam)

# Ringkasan
if len(beban_valid) > 0:
    rata = sum(beban_valid) / len(beban_valid)
    puncak = max(beban_valid)
    jam_puncak = jam_valid[beban_valid.index(puncak)]

    print("\n=== RINGKASAN ===")
    print(f"Data valid       : {len(beban_valid)} dari {len(data_beban)}")
    print(f"Sensor rusak     : {sensor_rusak}")
    print(f"Beban rata-rata  : {rata:.1f} kW")
    print(f"Beban puncak     : {puncak:.0f} kW (jam {jam_puncak:02d}:00)")

    # Simulasi ketahanan BBM
    konsumsi_per_jam = rata * KONSUMSI_LITER_PER_KWH
    sisa = TANGKI_LITER
    jam_tahan = 0
    while konsumsi_per_jam > 0 and sisa >= konsumsi_per_jam:
        sisa -= konsumsi_per_jam
        jam_tahan += 1

    print(f"Konsumsi BBM     : {konsumsi_per_jam:.1f} liter/jam")
    print(f"BBM {TANGKI_LITER} liter cukup untuk sekitar {jam_tahan} jam")

    # Rekomendasi
    if jumlah_kritis > 0:
        print(f"\nPERINGATAN: {jumlah_kritis} kali beban kritis, "
              "pertimbangkan menyalakan genset cadangan!")
    else:
        print("\nKondisi aman, tidak ada beban kritis.")
else:
    print("Tidak ada data valid untuk dianalisis.")
```
