```python
# Soal 1 — Kalkulator faktorial 1 sampai 5
n = int(input("Masukkan bilangan bulat (1–5): "))

# Validasi input: jika n kurang dari 1 ATAU lebih dari 5
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")

# Cabang untuk n == 1
elif n == 1:
    hasil = 1
    print("1! = 1 =", hasil)

# Cabang untuk n == 2
elif n == 2:
    hasil = 2 * 1
    print("2! = 2 x 1 =", hasil)

# Cabang untuk n == 3
elif n == 3:
    hasil = 3 * 2 * 1
    print("3! = 3 x 2 x 1 =", hasil)

# Cabang untuk n == 4
elif n == 4:
    hasil = 4 * 3 * 2 * 1
    print("4! = 4 x 3 x 2 x 1 =", hasil)

# Cabang terakhir: otomatis n == 5 (karena input sudah divalidasi 1-5)
else:
    hasil = 5 * 4 * 3 * 2 * 1
    print("5! = 5 x 4 x 3 x 2 x 1 =", hasil)
```

```python
# Soal 2 — Klasifikasi kualitas daya
faktor_daya = float(input("Masukkan faktor daya (0.0 – 1.0): "))
efisiensi   = float(input("Masukkan efisiensi distribusi (%): "))

# Hitung skor kualitas daya menggunakan operator aritmatika (*)
skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

# Periksa kondisi paling KETAT/SPESIFIK terlebih dahulu

if skor >= 85 and faktor_daya >= 0.90:   # skor >= 85 DAN faktor_daya >= 0.90
    klasifikasi = "PRIMA — sistem dalam kondisi optimal"

elif skor >= 70 and faktor_daya >= 0.80: # skor >= 70 DAN faktor_daya >= 0.80
    klasifikasi = "BAIK — sistem memadai, pantau rutin"

elif skor >= 50 or faktor_daya >= 0.75:  # skor >= 50 ATAU faktor_daya >= 0.75
    klasifikasi = "CUKUP — ada ruang perbaikan"

else:
    klasifikasi = "BURUK — perlu perbaikan segera"

print(f"Faktor daya {faktor_daya}, efisiensi {efisiensi}% → Klasifikasi: {klasifikasi}")
```

```python
# Soal 3 — Rekomendasi mode operasi PLTS
iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc      = float(input("Masukkan SoC baterai (%): "))

# Hitung estimasi daya panel menggunakan operator aritmatika
# Rumus: iradiasi * efisiensi_panel * luas_panel  (efisiensi = 0.18, luas = 10 m²)
daya_panel = iradiasi * 0.18 * 10

print(f"Estimasi daya panel: {daya_panel:.1f} W")

# Kondisi percabangan sesuai tabel

if iradiasi > 600 and soc < 80:    # iradiasi > 600 DAN SoC < 80
    mode = "CHARGING — prioritaskan pengisian baterai"

elif iradiasi > 600 and soc >= 80: # iradiasi > 600 DAN SoC >= 80
    mode = "GRID-EXPORT — ekspor kelebihan daya ke jaringan"

elif iradiasi > 200 and soc >= 20: # iradiasi > 200 DAN SoC >= 20
    mode = "NORMAL — operasi mandiri"

elif iradiasi <= 200 and soc >= 50: # iradiasi <= 200 DAN SoC >= 50
    mode = "BATTERY-ONLY — andalkan baterai"

else:
    mode = "EMERGENCY — kurangi beban segera"

print(f"Iradiasi {iradiasi} W/m², SoC {soc}% → Mode: {mode}")
```
